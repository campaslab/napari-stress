# Surface reconstruction

This notebook demonstrates how to reconstruct a surface from a pointcloud with napari-stress. This plugin uses the `recoSurface` function as implemented in [vedo](https://vedo.embl.es/autodocs/content/vedo/pointcloud.html#vedo.pointcloud.Points.reconstructSurface).

```python
import napari
import vedo

import napari_stress
```

```python
viewer = napari.Viewer(ndisplay=3)
```

Let's create some sample data:

```python
points = vedo.shapes.Ellipsoid(res=100).vertices * 10
points_layer = viewer.add_points(points, size=0.5)
napari.utils.nbscreenshot(viewer)
```

```python
surface = napari_stress.reconstruct_surface(points, radius=5)
```

```python
points_layer.visible = False
viewer.add_surface(surface)
napari.utils.nbscreenshot(viewer)
```
